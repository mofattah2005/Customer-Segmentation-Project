```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# قراءة الداتا من الرابط المباشر
df = pd.read_csv(r'C:\Users\workstation\Desktop\Project Data Science\online_retail.csv')

print("تم تحميل البيانات بنجاح، وعدد الصفوف هو:", df.shape)
display(df.head())

# تنظيف البيانات من القيم المفقودة للعملاء
df = df.dropna(subset=['CustomerID'])

# تحويل التاريخ وحساب إجمالي المشتريات
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['TotalSum'] = df['Quantity'] * df['UnitPrice']

# تحديد التاريخ المرجعي لحساب الحداثة
snapshot_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)

# حساب مؤشرات RFM لكل عميل
rfm_df = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,
    'InvoiceNo': 'nunique',
    'TotalSum': 'sum'
})

rfm_df.columns = ['Recency', 'Frequency', 'Monetary']
display(rfm_df.head())

# توحيد مقاييس البيانات
scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm_df[['Recency', 'Frequency', 'Monetary']])

# تطبيق K-Means باختيار 4 مجموعات
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans.fit(rfm_scaled)

# إضافة رقم المجموعة للجدول
rfm_df['Cluster'] = kmeans.labels_
display(rfm_df.head(10))

# قاموس لترجمة أرقام المجموعات إلى أسماء تسويقية واضحة
cluster_names = {
    2: 'VIP Champions',
    3: 'Loyal Customers',
    0: 'Regular Customers',
    1: 'At-Risk / Inactive'
}


# استبدال الأرقام بالأسماء في جدول الـ RFM
rfm_df['Cluster_Name'] = rfm_df['Cluster'].map(cluster_names)

# عرض عينة للتأكد من النتيجة
display(rfm_df[['Recency', 'Frequency', 'Monetary', 'Cluster_Name']].head(10))

# حساب متوسطات المجموعات لعرضها في التقرير
cluster_summary = rfm_df.groupby('Cluster').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': ['mean', 'count']
}).round(2)

print("ملخص خصائص كل مجموعة:")
display(cluster_summary)

# رسم بياني يوضح توزيع العملاء
plt.figure(figsize=(8, 5))
sns.scatterplot(data=rfm_df, x='Recency', y='Monetary', hue='Cluster_Name', palette='Set1')
plt.title('Customer Segments based on Recency and Monetary')
plt.xlabel('Recency (Days)')
plt.ylabel('Monetary ($)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') # لضبط مكان المفتاح بره الرسمة عشان يكون واضح
plt.show()
```
